```python
import pandas as pd
from sklearn.model_selection import train_test_split
from google.colab import files

# Upload the CSV from your local computer
uploaded = files.upload()

# Read the uploaded file
df = pd.read_csv("creditcard.csv")

print("Original dataset:", df.shape)
print(df["Class"].value_counts())

# 10% final unseen validation set
development_data, validation_data = train_test_split(
    df,
    test_size=0.10,
    stratify=df["Class"],
    random_state=42
)

# Split remaining 90% into 70% training and 20% test
training_data, test_data = train_test_split(
    development_data,
    test_size=2/9,
    stratify=development_data["Class"],
    random_state=42
)

print("\nTraining Dataset:", training_data.shape)
print("Test Dataset:", test_data.shape)
print("Validation Dataset:", validation_data.shape)

print("\nTraining Class Distribution:")
print(training_data["Class"].value_counts())

print("\nTest Class Distribution:")
print(test_data["Class"].value_counts())

print("\nValidation Class Distribution:")
print(validation_data["Class"].value_counts())

# Save the three datasets
training_data.to_csv("training_data.csv", index=False)
test_data.to_csv("test_data.csv", index=False)
validation_data.to_csv("validation_data.csv", index=False)

print("\nThree datasets successfully created.")
```
