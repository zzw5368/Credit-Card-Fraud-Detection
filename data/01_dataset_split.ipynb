```python
import pandas as pd
from sklearn.model_selection import train_test_split
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')
```

```python
# Path to the original dataset in Google Drive
file_path = "/content/drive/MyDrive/DS340W/creditcard.csv"

# Load dataset
df = pd.read_csv(file_path)

print("Original dataset:", df.shape)
print(df["Class"].value_counts())
```

```python
# -------------------------------------------------
# STEP 1: Separate 10% as final unseen validation data
# -------------------------------------------------

development_data, validation_data = train_test_split(
    df,
    test_size=0.10,
    stratify=df["Class"],
    random_state=42
)

# -------------------------------------------------
# STEP 2: Split remaining 90% into 70% training
# and 20% testing
# -------------------------------------------------

training_data, test_data = train_test_split(
    development_data,
    test_size=2/9,
    stratify=development_data["Class"],
    random_state=42
)

# Check sizes
print("\nTraining Dataset:", training_data.shape)
print("Test Dataset:", test_data.shape)
print("Validation Dataset:", validation_data.shape)

# Check class distribution
print("\nTraining Class Distribution:")
print(training_data["Class"].value_counts())

print("\nTest Class Distribution:")
print(test_data["Class"].value_counts())

print("\nValidation Class Distribution:")
print(validation_data["Class"].value_counts())
```

```python

output_folder = "/content/drive/MyDrive/DS340W/"

training_data.to_csv(
    output_folder + "training_data.csv",
    index=False
)

test_data.to_csv(
    output_folder + "test_data.csv",
    index=False
)

validation_data.to_csv(
    output_folder + "validation_data.csv",
    index=False
)

print("\nThree datasets successfully saved to Google Drive.")
```
