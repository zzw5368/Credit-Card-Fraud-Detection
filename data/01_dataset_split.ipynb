# Credit Card Fraud Detection
## Baseline vs Improved Ensemble

**Baseline:** AdaBoost + Random Forest + KNN

**Improved:** Bayesian-optimized XGBoost + Random Forest + KNN

Both ensembles use SMOTE + ENN on training data only. The experiment compares accuracy, precision, recall, F1, ROC-AUC, PR-AUC, and confusion matrices.
